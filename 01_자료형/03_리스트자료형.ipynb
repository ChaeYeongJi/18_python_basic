```python
# 리스트는 어떻게 만들고 사용할까?
# 리스트명 = [요소1, 요소2, ...]

a = []
b = [1, 2, 3]
c = ["Life", "is", "too", "short"]
d = [1, 2, ["life", "is"]]
e = [1, 2, "life", "is"]
f = list()

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```

```python
## 리스트의 인덱싱
a = [1, 2, 3]
type(a)
a[1]
a[-len(a)]
```

```python
a = [1, 2, 3, ["a", "b", "c"]]
len(a)
a[3][2]
```

```python
## 리스트의 슬라이싱
a = [1, 2, 3, 4, 5]
a[1:-1]
```

```python
a = [1, 2, 3, ["a", "b", "c"]]
a[-1][1:]
```

```python
## 리스트 연산
a = [1, 2, 3]
b = [4, 5, 6]

a+b
a*3
```

```python
## 리스트 길이 구하기 - len
a = [1, 2, 3]
len(a)
```

```python
## 리스트의 수정과 삭제
a = [1, 2, 3]
a[2] = 4
a
```

```python
a = [1, 2, 3]
del a[1]
print(a)
```

```python
## 리스트 관련 함수

## 리스트에 요소 추가하기 - append
a = [1, 2, 3]
a.append(4)
print(a)

a.append(["a", "b"])
print(a)
```

```python
## 리스트 정렬 - sort
a = [1, 4, 3, 2]
a.sort()
print(a)

a = ["a", "c", "b"]
a.sort(reverse=True)
print(a)
```

```python
## 리스트 뒤집기 - reverse
a = ["a", "c", "b"]
a.reverse()
a
```

```python
## 인덱스 반환 - index
a = [11, 2, 23, 4]
a.index(23)
```

```python
## 리스트에 요소 삽입 - insert
a = [1, 2, 3]
a.insert(-1, 22)
print(a)
a.sort()
print(a)
```

```python
## 리스트에 요소 제거 - remove()
a = [1, 2, 3, 1, 2, 3]
a.remove(2)
print(a)
```

```python
## 리스트 요소 꺼내기 - pop
a = [1, 2, 3]
number = a.pop(-2)
print(number)
print(a)
```

```python
## 리스트에 포함된 요소 x의 개수 세기 - count
a = [1, 2, 3, 4, 3, 3, 3, 4]
a.count(3)
```

```python
## 리스트 확장
a = [1, 2, 3]
a.extend([5])
a
```
