```python
# if 조건문의 동작 방식 알아보기
if True:
    print('참')
else:
    print('거짓')

if False:
    print('참')
else:
    print('거짓')

if None:
    print('참')
else:
    print('거짓')
```

```python
# if 조건문에 숫자 지정하기

if 0:
    print('참')
else:
    print('거짓')

if 1:
    print('참')
else:
    print('거짓')

if 0x1f:
    print('참')
else:
    print('거짓')

if 0b1000:
    print('참')
else:
    print('거짓')

if 13.5:
    print('참')
else:
    print('거짓')
```

```python
# if 조건문에 문자열 지정하기
if 'hello':
    print('참')
else:
    print('거짓')

if '':
    print('참')
else:
    print('거짓')

if ' ':
    print('참')
else:
    print('거짓')
```

```python
# 조건식을 여러개 지정하기
x = 10
y = 20

if x == 10 and y == 20:
    print('참')
else:
    print('거짓')
```

```python
# 중첩 if 조건문과 논리 연산자
x = 10

if x > 0:
    if x < 20:
        print('20보다 작은 양수입니다.')

if x > 0 and x < 20:
    print('20보다 작은 양수입니다.')

if 0 < x < 20:
    print('20보다 작은 양수입니다.')
```

```python
# c / b, d / True / b, e / a

# 연습문제
# written_test = 75
# coding_test = True

# if written_test >= 80 and coding_test:
#     print('합격')
# else:
#     print('불합격')


# 심사문제
# 국어, 영어, 수학, 과학 평균 80점 이상일 때 합격. 점수 범위는 0~100점. 범위 외의 경우 '잘못된 점수' 출력
kor, eng, math, sci = map(int, input().split())
if 0 <= kor <= 100 and 0 <= eng <= 100 and 0 <= math <= 100 and 0 <= sci <= 100:
    if ((kor + eng + math + sci) / 4) >= 80:
        print('합격')
    else:
        print('불합격')
else:
    print('잘못된 점수')
```
