```python
# while 반복문 사용하기

# 초기식
# while 조건식:
#     반복할 코드
#     변화식

i = 0
while i < 10:
    print(i, 'hello world')
    i += 1
```

```python
# 초깃값을 1부터 시작하기
i = 1
while i <= 5:
    print('hello world', i)
    i += 1
```

```python
# 초깃값을 감소시키기
i = 5
while i > 0:
    print(i)
    i -= 1
```

```python
count = int(input())

i = 0
while i < count:
    print(i)
    i += 1
```

```python
# 반복 횟수가 정해지지 않은 경우
import random

print(int(random.random() * 10))
print(random.randint(1, 6))
```

```python
import random

i = 0
while i != 3:
    i = random.randint(1, 6)
    print(i)
```
