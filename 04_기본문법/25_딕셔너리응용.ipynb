```python
# 딕셔너리 조작하기
x = {
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
}

x.setdefault('e')
print(x)

x.setdefault('f', 100)
print(x)
```

```python
# 딕셔너리에서 키의 값 수정하기
x = {
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
}

# 기존 값 수정
x.update(a=90) # x['a'] = 90 과 동일
print(x)

# 없던 값도 추가 가능
x.update(e=50)
print(x)
```

```python
x.update(a=900, f=60)
print(x)
```

```python
y = {
    1: 'one',
    2: 'two'
}
print(y)

y.update({
    1: 'ONE',
    2: 'TWO'
})
print(y)

y.update(
    [
        [2, 'TWOTWO'],
        [4, 'FOUR']
    ]
)
print(y)

y.update(
    zip([1, 2], ['re-one', 're-two'])
)
print(y)
```

```python
# 딕셔너리에서 키-값 쌍 삭제하기
x = {
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
}

print(x.pop('a'))
print(x)

# 없는 키를 입력했을 때, 오류처리 0으로
print(x.pop('z', 0))
print(x)

del x['b']
print(x)
```

```python
# 딕셔너리에서 임의의 키-값 쌍 삭제하기
x = {
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
}

print(x.popitem())
print(x)

# 딕셔너리에서 모든 키-값 쌍을 삭제하기
x.clear()
print(x)
```

```python
# 딕셔너리에서 키의 값을 가져오기
x = {
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
}

print(x['a'])
# print(x['z'])

print(x.get('a'))
print(x.get('z'))
print(x.get('z', 0))
```

```python
# 딕셔너리에서 키-값 쌍을 모두 가져오기
x = {
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
}

print(type(x.items()), x.items())
print(list(x.items()))

# keys() => 모든 키를 가져옵니다.
print(type(x.keys()), x.keys())
print(list(x.keys()))

# values() => 모든 값을 가져옵니다.
print(type(x.values()), x.values())
print(list(x.values()))
```

```python
# 리스트와 튜플로 딕셔너리 만들기
keys = ['a', 'b', 'c', 'd']
x = dict.fromkeys(keys)
print(x)

print({}.fromkeys(keys))

y = dict.fromkeys(keys, 100)
print(y)
```
