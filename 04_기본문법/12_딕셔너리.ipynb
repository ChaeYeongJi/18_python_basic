```python
# 딕셔너리 만들기
# 딕셔너리 = {
#     키1: 값1,
#     키2: 값2
# }

lux = {
    'health': 490,
    'mana': 334,
    'melee': 550,
    'armor': 18.72
}
print(lux)
lux['health']
```

```python
# 빈 딕셔너리 만들기

딕셔너리 = {}
딕셔너리 = dict()

# int(), str(), float(), list(), tuple(), bool()

x = {}
y = dict()

print(type(x), x)
print(type(y), y)
```

```python
# dict로 딕셔너리 만들기
lux1 = dict(health = 490, mana = 334, melee = 550, armor = 18.72)
print(lux1)
```

```python
iter1 = ['health', 'mana', 'melee', 'armor']
iter2 = [490, 334, 550, 18.72]
# list(zip(iter1, iter2))
dict(zip(iter1, iter2))
```

```python
lux3 = dict(
    [("health", 490), ('mana', 334), ('melee', 550), ('armor', 18.72)]
)
lux3
```

```python
lux4 = dict(
    {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
)
lux4
```

```python
# 딕셔너리의 키에 접근하고 값 할당하기
lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
print(lux['health'])
print(lux['armor'])
```

```python
# 딕셔너리의 키에 값 할당하기
lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
lux['health'] = 20237
lux['mana'] = 1184

print(lux)

lux['mana_regen'] = 3.28

print(lux)

# lux['attack_speed']
```

```python
# 딕셔너리에 키가 있는지 확인하기
lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
'health' in lux
```

```python
# 딕셔너리의 키 개수 구하기
lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
len(lux)
```
