```python
# 2차원 리스트의 요소에 접근하기
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

print(a[0])
print(a[0][0])
print(a[0][1])
print(a[1])
print(a[1][0])

a[0][1] = 1000

print(a)
```

```python
# 반복문으로 2차원 리스트의 요소를 모두 출력하기
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

for i in a:
    for j in i:
        print(j, end=' ')
    print()

# for x, y in a:
#     print(x, y, end=' ')
```

```python
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

for i in range(len(a)):
    for j in range(len(a[i])):
        print(a[i][j], end=' ')
    print()
```

```python
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

i = 0
while i < len(a):
    x, y = a[i]
    print(x, y)
    i += 1
```

```python
# while 반복문을 두 번 사용하기
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

i = 0
while i < len(a):
    j = 0
    while j < len(a[i]):
        print(a[i][j], end=' ')
        j += 1
    print()
    i += 1
```

```python
# for 반복문으로 1차원 리스트 만들기
a = []

for _ in range(10):
    a.append(0)

print(a)
```

```python
# for 반복문으로 2차원 리스트 만들기
a = []
for _ in range(3):
    line = []
    for _ in range(2):
        line.append(0)
    a.append(line)

print(a)
```

```python
# 톱니형 리스트 만들기
a = [3, 1, 3, 2, 5]
b = []

for i in a:
    line = []
    for j in range(i):
        line.append(0)
        # 1 [0, 0, 0]
        # 2 [0]
        # 3 [0, 0, 0]
        # 4 [0, 0]
        # 5 [0, 0, 0, 0, 0]
    b.append(line)

print(b)
```

```python
col, row = map(int, input().split())
matrix = []
for i in range(row):
    matrix.append(list(input()))

for i in range(row):
    for j in range(col):
        if matrix[i][j] == '.':
            count = 0
            for r in range(i-1, i+2):
                for c in range(j-1, j+2):
                    if 0 <= r < row and 0 <= c < col:
                        if matrix[r][c] == '*':
                            count += 1
            matrix[i][j] = count

print(*matrix, sep='\n')
```
