```python
# 2차원 리스트 만들고 요소에 접근하기

a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

print(a)
```

```python
# 2차원 리스트의 요소에 접근하기
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

print(a[0])
print(a[0][0])
print(a[0][1])
print(a[1])
print(a[1][0])

a[0][1] = 1000

print(a)
```

```python
# 반복문으로 2차원 리스트의 요소를 모두 출력하기
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

for i in a:
    for j in i:
        print(j, end=' ')
    print()

# for x, y in a:
#     print(x, y, end=' ')
```

```python
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

for i in range(len(a)):
    for j in range(len(a[i])):
        print(a[i][j], end=' ')
    print()
```

```python
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

i = 0
while i < len(a):
    x, y = a[i]
    print(x, y)
    i += 1
```

```python
# while 반복문을 두 번 사용하기
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

i = 0
while i < len(a):
    j = 0
    while j < len(a[i]):
        print(a[i][j], end=' ')
        j += 1
    print()
    i += 1
```
