```python
# 1부터 100까지 숫자 출력하기
for i in range(1, 21):
    if i % 3 == 0 and i % 5 == 0:
        print('FizzBuzz')
    elif i % 3 == 0:
        print('Fizz')
    elif i % 5 == 0:
        print('Buzz')
    else:
        print(i)
```

```python
# 코드 단축하기

for i in range(11, 31):
    print('Fizz' * (i % 3 == 0) + 'Buzz' * (i % 5 == 0) or i)
```
