```python
# 문자열 포매팅에 변수를 그대로 사용하기
language = 'Python'
version = 3.6

print(f'Hello {language} {version}')
```

```python
# format 메서드로 문자열 정렬하기
# "{인덱스:<10}".format('python')
print("{0:<10}".format('python'))
print("{0:>10}".format('python'))
print("{:>10}".format('python'))
```

```python
# 숫자 개수 맞추기
# "%0개수d" % 숫자
# "{인덱스:0개수d}".format(숫자)
print("%d" % 1)
print("%03d" % 1)
print("{0}".format(35))
print("{1:03d}".format(35, 33))
```

```python
print("%08.2f" % 3.6)
print('{:08.2f}'.format(150.37))
```

```python
# 채우기와 정렬을 조합해서 사용하기
print("{0}".format(15))
print("{:<10}".format(15))
print("{:>10}".format(15))
```

```python
number = 3.6
print(f"{number:010.2f}")
```

```python
number = 30000
print(f"{number:,.2f}")
```
