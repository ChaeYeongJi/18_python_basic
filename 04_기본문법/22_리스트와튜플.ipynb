```python
# 리스트의 할당과 복사하기
a = [0, 0, 0, 0, 0]
b = a
b[2] = 99
a is b
print(id(a), a)
print(id(b), b)

a = [0, 0, 0, 0, 0]
b = a.copy()
print(a is b)
print(a == b)
print(id(a), a)
print(id(b), b)
b[2] = 99
print(a)
print(b)
```

```python
# 리스트 표현식 사용하기(리스트 컴프리헨션)
## 1. [식 for 변수 in 리스트]
## 2. list(식 for 변수 in 리스트)
```

```python
a = []
for i in range(10):
    a.append(i)

print(a)
```

```python
a = [i for i in range(10)]
print(a)
```

```python
b = list(i for i in range(10))
print(b)
```

```python
c = [i + 5 for i in range(10)]
print(c)
```

```python
d = [i * 2 for i in range(10)]
print(d)
```

```python
# 리스트 표현식에서 if 조건문 사용하기
# a = []
# for i in range(10):
#     if i % 2 == 0:
#         a.append(i)
# print(a)

a = [i for i in range(10) if i % 2 == 0]
print(a)
```

```python
# b = [i + 5 for i in range(10) if i % 2 == 1]
# print(b)

b = []
for i in range(10):
    if i % 2 == 1:
        b.append(i + 5)
print(b)
```

```python
# for 반복문과 if 조건문을 여러 번 사용하기
[i * j for i in range(2, 10) for j in range(1, 10)]
```
